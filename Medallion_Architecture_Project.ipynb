```python
bronze_data = [
    (1, "Jaipur", "Electronics", 500),
    (2, "Delhi", "Clothing", 700),
    (2, "Delhi", "Clothing", 700),
    (3, "Mumbai", "Electronics", 900),
    (4, "Jaipur", "Furniture", 1200),
    (5, "Delhi", "Electronics", 1500)
]

bronze_df = spark.createDataFrame(
    bronze_data,
    ["OrderID", "City", "Category", "Amount"]
)

display(bronze_df)
```

```python
print("Rows:", bronze_df.count())
print("Columns:", len(bronze_df.columns))
```

```python
silver_df = bronze_df.dropDuplicates()

display(silver_df)
```

```python
silver_df = silver_df.na.fill(0)

display(silver_df)
```

```python
from pyspark.sql.functions import sum

gold_df = silver_df.groupBy("City") \
                   .agg(sum("Amount").alias("TotalSales"))

display(gold_df)
```

```python
category_df = silver_df.groupBy("Category") \
                       .agg(sum("Amount").alias("CategorySales"))

display(category_df)
```

```python
display(
    gold_df.orderBy("TotalSales", ascending=False)
)
```

```python
gold_df.write.mode("overwrite").saveAsTable("gold_sales")
```
